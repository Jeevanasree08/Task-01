```python
# ----------------------------
# A. Hierarchical Indexing
# ----------------------------

# Create a Series with random values and a hierarchical (multi-level) index
data = pd.Series(
    np.random.randn(6),
    index=[['a', 'a', 'b', 'b', 'c', 'c'],
           [1, 2, 1, 2, 1, 2]]
)
```

```python
# Display the complete Series
print('Original Series:')
print(data)
```

```python
# Display all values belonging to index 'b'
print('\nValues for index b:')
print(data['b'])
```

```python
# Display values where second-level index equals 1
print('\nValues with second index = 1:')
print(data[:, 1])
```

```python
# ----------------------------
# B. Stack and Unstack
# ----------------------------

# Convert hierarchical Series into DataFrame
df = data.unstack()
```

```python
# Display unstacked DataFrame
print('\nUnstacked DataFrame:')
print(df)
```

```python
# Convert DataFrame back to Series
print('\nStacked Series:')
print(df.stack())
```

```python
# ----------------------------
# C. Merge and Combine
# ----------------------------

# Create first DataFrame with missing values
df1 = pd.DataFrame({'a': [1, np.nan, 3], 'b': [4, 5, 6]})
```

```python
# Create second DataFrame with missing values
df2 = pd.DataFrame({'a': [7, 8, 9], 'b': [np.nan, 11, 12]})
```

```python
# Combine DataFrames and fill missing values in df1 using df2
merged = df1.combine_first(df2)
```

```python
# Display merged DataFrame
print('\nMerged DataFrame:')
print(merged)
```

```python
# DATA SCIENCE - Program 6
# Time Series Analysis
# Import pandas library for time series operations
import pandas as pd
```

```python
#Create a date range starting from 2023-01-01
# periods=6 creates 6 dates
# freq='D' means daily frequency
rng = pd.date_range('2023-01-01', periods=6, freq='D')
```

```python
"""Program 8"""

# DATA SCIENCE - PROGRAM 8
# TOPIC: WEB SCRAPING USING BEAUTIFUL SOUP
# requests library is used to send HTTP requests
# and download the web page content.
import requests
```

```python
# BeautifulSoup is used to parse HTML/XML documents
# and extract useful information from web pages.
from bs4 import BeautifulSoup
```

```python
# Many websites block requests coming from bots.
# To make our request look like it is coming from
# a real browser, we provide a User-Agent header.
#
# User-Agent identifies the browser and operating system.
headers = {
    "User-Agent": "Mozilla/5.0"
}
```

```python
# Amazon review page URL.
# This page contains customer reviews for a product.
url = "https://www.amazon.in/product-reviews/B09G9BL5CP/"
```

```python
# Send a GET request to the website.
# The server returns the HTML source code.
r = requests.get(url, headers=headers)
```

```python
# Print the HTTP status code.
# 200 = Success
# 404 = Not Found
# 500 = Server Error
print("Status Code:", r.status_code)
```

```python
# Convert downloaded HTML content into a BeautifulSoup object.
# html.parser is Python's built-in HTML parser.
soup = BeautifulSoup(r.content, 'html.parser')
```

```python
# Amazon stores review text inside span tags
# with the attribute:
# data-hook="review-body"
#
# find_all() returns a list of matching elements.
reviews = soup.find_all(
    'span',
    {'data-hook': 'review-body'}
)
```

```python
# Print total reviews found on the page.
print("Number of Reviews Found:", len(reviews))
```

```python
# Select the first 5 reviews
for review in reviews[:5]:
    print(review)
```

```python
for review in reviews[:5]:

    # Extract only the review text.
    # strip() removes extra spaces and new lines.
    review_text = review.text.strip()

    # Print review text.
    print("\nReview:")
    print(review_text)

print("\nScraping Completed Successfully")
```

```python

# ----
# To analyze customer purchasing behavior and divide customers
# into different groups (clusters) using the K-Means clustering
# algorithm.
#
# K-Means is an Unsupervised Machine Learning algorithm.
# It groups similar customers together without predefined labels.
#
# ==============================================================


# ==============================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ==============================================================

import pandas as pd
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

# ==============================================================
# STEP 2: LOAD THE DATASET
# ==============================================================

# IMPORTANT:
# Use the RAW GitHub URL, not the GitHub /blob/ webpage URL.

url = "https://raw.githubusercontent.com/tirthajyoti/Machine-Learning-with-Python/master/Datasets/Mall_Customers.csv"

```

```python
# Read CSV file
df = pd.read_csv(url)

```

```python
# ==============================================================
# STEP 6: SELECT FEATURES FOR CLUSTERING
# ==============================================================

# We use:
#
# Annual Income (k$)
# Spending Score (1-100)
#
# These two variables are commonly used for customer segmentation.

X = df[
    [
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
]
```

```python
print("\n")
print("=" * 70)
```

```python
print("SELECTED FEATURES")
print("=" * 70)
```

```python
print(X.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Standardize the features
scaler = StandardScaler()

```

```python
X_scaled = scaler.fit_transform(X)
```

```python
print("\n")
print("=" * 70)
```

```python
print("STANDARDIZED FEATURES")
print("=" * 70)
```

```python
print(X_scaled[:5])

```

```python
from sklearn.cluster import KMeans
```

```python
inertia = []

for k in range(1, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    inertia.append(model.inertia_)
```

```python
# Plot Elbow Curve
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inertia,
    marker="o"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.grid(True)

plt.tight_layout()

plt.savefig("elbow_method.png")

plt.show()
```

```python
# ==============================================================
# STEP 9: CREATE K-MEANS MODEL
# ==============================================================

# Based on the elbow method, we select 5 clusters.

kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

```

```python
# ==============================================================
# STEP 10: TRAIN MODEL AND ASSIGN CLUSTERS
# ==============================================================

# fit_predict():
#
# fit()         -> trains the K-Means model
# predict()     -> assigns cluster numbers
#
# fit_predict() performs both operations.

df["Cluster"] = kmeans.fit_predict(X_scaled)

```

```python
# ==============================================================
# STEP 11: DISPLAY CUSTOMER CLUSTERS
# ==============================================================

print("\n")
print("=" * 70)
print("CUSTOMER CLUSTERS")
print("=" * 70)

```

```python
print(
    df[
        [
            "CustomerID",
            "Annual Income (k$)",
            "Spending Score (1-100)",
            "Cluster"
        ]
    ].head(20)
)
```

```python
# ==============================================================
# STEP 12: DISPLAY CLUSTER CENTERS
# ==============================================================

# Cluster centers are initially calculated
# in standardized feature space.
#
# Convert them back to the original scale
# for easier interpretation.

cluster_centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)
```

```python
centers_df = pd.DataFrame(
    cluster_centers,
    columns=[
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
)

centers_df.index.name = "Cluster"


```

```python
print("\n")
print("=" * 70)
print("CLUSTER CENTERS")
print("=" * 70)

print(centers_df)
```

```python

# ==============================================================
# STEP 13: COUNT CUSTOMERS IN EACH CLUSTER
# ==============================================================

print("\n")
print("=" * 70)
print("CUSTOMER COUNT PER CLUSTER")
print("=" * 70)

print(df["Cluster"].value_counts().sort_index())

```

```python
# ==============================================================
# STEP 14: CUSTOMER SEGMENT SUMMARY
# ==============================================================

summary = df.groupby("Cluster")[
    [
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
].mean()

summary["Customer Count"] = df["Cluster"].value_counts()

summary = summary.sort_index()


print("\n")
print("=" * 70)
print("CUSTOMER SEGMENT SUMMARY")
print("=" * 70)

print(summary)
```

```python
# ==============================================================
# STEP 15: VISUALIZE CUSTOMER SEGMENTS
# ==============================================================

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="tab10",
    s=100
)


# Plot cluster centers

plt.scatter(
    cluster_centers[:, 0],
    cluster_centers[:, 1],
    marker="X",
    s=250,
    c="black",
    label="Cluster Centers"
)


plt.title(
    "Customer Personality Analysis using K-Means",
    fontsize=15
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "customer_segments_kmeans.png",
    dpi=300
)

plt.show()
```

```python
# ==============================================================
# STEP 16: SAVE CLUSTERED DATASET
# ==============================================================

df.to_csv(
    "Mall_Customers_Clustered.csv",
    index=False
)
```

```python
# ==============================================================
# STEP 17: FINAL MESSAGE
# ==============================================================

print("\n")
print("=" * 70)
print("CUSTOMER SEGMENTATION COMPLETED SUCCESSFULLY")
print("=" * 70)

print("Number of Clusters :", kmeans.n_clusters)

print("Clustered dataset saved as:")
print("Mall_Customers_Clustered.csv")

print("\nGraphs saved as:")
print("1. elbow_method.png")
print("2. customer_segments_kmeans.png")
```

```python
# ============================================================
# DATA SCIENCE - PROGRAM 10
# CASE STUDY: TEXT EMOTION DETECTION
# ============================================================
#
# AIM:
# ----
# To detect emotions from text messages using
# Machine Learning techniques.
#
# In this program we:
#
# 1. Create sample text data.
# 2. Convert text into numerical features.
# 3. Train a Naive Bayes classifier.
# 4. Predict emotion for new text.
#
# ============================================================

# ============================================================
# IMPORT REQUIRED LIBRARIES
# ============================================================

# CountVectorizer converts text into numerical vectors.
# Machine learning algorithms cannot understand text directly.
# Therefore, text must be converted into numbers.
from sklearn.feature_extraction.text import CountVectorizer

```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
# ============================================================
# CREATE SAMPLE TRAINING DATA
# ============================================================

# List of sample text messages.
texts = [
    "I am happy",
    "I am angry",
    "I feel sad",
    "I love this"
]
```

```python
# Corresponding labels (emotions).
labels = [
    "happy",
    "angry",
    "sad",
    "happy"
]
```

```python
# Display training data.
print("=" * 60)
print("TRAINING DATA")
print("=" * 60)
```

```python
for sentence, emotion in zip(texts, labels):
    print(f"Text: {sentence} --> Emotion: {emotion}")
```

```python
# depending on the vocabulary.
vec = CountVectorizer()
```

```python
# Learn vocabulary from training data.
# Convert text into numerical matrix.
X = vec.fit_transform(texts)
```

```python

# Display all learned words.
print("\n")
print("=" * 60)
print("VOCABULARY")
print("=" * 60)

```

```python
# Display numerical representation.
print("\n")
print("=" * 60)
print("VECTORIZED DATA")
print("=" * 60)
```

```python
print(X.toarray())
```

```python
# ============================================================
# TRAIN MACHINE LEARNING MODEL
# ============================================================

# Create Multinomial Naive Bayes model.
model = MultinomialNB()
```

```python

# Train the model.
#
# X      -> Features
# labels -> Target classes
model.fit(X, labels)

print("\nModel Training Completed Successfully")
```

```python
# ============================================================
# TEST DATA
# ============================================================

# New sentence whose emotion must be predicted.
new_text = ["I am so sad today"]

```

```python
print("\n")
print("=" * 60)
print("TEST SENTENCE")
print("=" * 60)

print(new_text)
```

```python
# Convert test text into numerical representation
# using the vocabulary learned earlier.
test = vec.transform(new_text)

```

```python
# Display vectorized form of test data.
print("\nVectorized Test Data:")
print(test.toarray())
```

```python
# Predict emotion.
prediction = model.predict(test)

```

```python
# Display prediction.
print("\n")
print("=" * 60)
print("PREDICTED EMOTION")
print("=" * 60)

```

```python
print(prediction[0])
```

```python
# ============================================================
# ADDITIONAL TEST CASES
# ============================================================

sample_sentences = [
    "I love my friends",
    "I am angry with you",
    "I feel very sad",
    "Today I am happy"
]

```

```python
print("\n")
print("=" * 60)
print("MULTIPLE PREDICTIONS")
print("=" * 60)
```

```python
for sentence in sample_sentences:

    # Convert text into vector
    sample_vector = vec.transform([sentence])

    # Predict emotion
    result = model.predict(sample_vector)

    print(f"{sentence} --> {result[0]}")
```

```python
# ============================================================
# END OF PROGRAM
# ============================================================

print("\n")
print("=" * 60)
print("TEXT EMOTION DETECTION COMPLETED")
print("=" * 60)
```
